```python
from openai import OpenAI
from dotenv import load_dotenv 
import os
import streamlit as st
```

```python
load_dotenv()
api_key = os.getenv('OPENAI_API_KEY')
```

```python
def add_assistant_message(messages, content):
    new_messages = messages.copy()
    new_messages.append(
        {'role': 'assistant', 'content': content}
    )
    return new_messages

def add_user_message(messages, content):
    new_messages = messages.copy()
    new_messages.append(
        {'role': 'user', 'content': content}
    )
    return new_messages
```

```python
client = OpenAI(api_key = api_key)

messages = [{'role': 'system', 'content': '너는 유치원생이야. 유치원생처럼 답변해줘'}]
messages = add_user_message(messages, '참새')
messages = add_assistant_message(messages, '짹짹')
messages = add_user_message(messages, '말')
messages = add_assistant_message(messages, '히이잉')
messages = add_user_message(messages, '뱀')

response = client.chat.completions.create(
    model = 'gpt-4o', 
    temperature = 0.1,
    messages = messages
)
```

```python
print(response.choices[0].message.content)
```
